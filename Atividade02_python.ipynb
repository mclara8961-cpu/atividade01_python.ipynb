```python
# 4-Estacionamento

x = int(input('Digite a quantidade de hora:'))

if x <= 2:
  v = 8
elif x <= 5:
  v = 8+(x-2)*3
else:
  v = 17+(x-5)*2

print(f'O valor a ser pago é R$:{v}.')
```

```python
# 5-Máquina de venda de ingressos

t = 0

while t < 25:
    v = int(input('Digite o valor inserido: '))
    t = t + v

s = t - 25

print(f'O valor digitado e o troco são respectivamente {t} e {s}.')
```

```python
# 6-Maior valor e sua posição

p = int(input("Digite o número: "))

m = p
posicao = 1

for i in range(2, 9):
    numero = int(input("Digite o número: "))

    if numero > m:
        m = numero
        posicao = i

print('Maior valor:', m)
print('Posição:', posicao)
```

```python
# 7-Descobrindo o segundo maior valor

p = int(input('Digite o número: '))
s = int(input('Digite o número: '))

if p > s:
    m = p
    sm = s
else:
    m = s
    sm = p

for i in range(6):
    n = int(input("Digite o número: "))

    if n > m:
        sm = m
        m = n

    elif n > sm:
        sm = n

print("Maior:", m)
print("Segundo maior:", sm)
```

```python
# 8-Detectando valores consecutivos

ant = int(input("Digite o primeiro número: "))

rep = False

for i in range(7):
    at = int(input("Digite o número: "))

    if at == ant:
        rep = True

    ant = at

if rep:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

```python
# 09-Controle de saldo

s = 500

for i in range(6):
    o = int(input('1-Depósito | 2-Saque: '))
    v = float(input('Digite o valor: '))

    if o == 1:
        s = s + v

    elif o == 2:
        if v <= s:
            s = s - v
        else:
            print("Saldo insuficiente")

print("Saldo final: R$", s)
```

```python
# 12-Divisores de N

n = int(input("Digite um número: "))

d = 0

for i in range(1, n + 1):
    if n % i == 0:
        print(i)
        d = d + 1

print("Quantidade de divisores:", d)
```

```python
# 13-Números primos em um intervalo

a = int(input('Digite o valor de A: '))
b = int(input('Digite o valor de B: '))

q = 0

for n in range(a, b + 1):
    primo = True

    if n < 2:
        primo = False

    for i in range(2, n):
        if n % i == 0:
            primo = False

    if primo:
        print(n)
        q = q + 1

print(f'Quantidade de primos {q}.')
```

```python
# 14-Jogo com limites de tentativa

n = 67
ac = False

for t in range(5):
    t = int(input('Digite o número: '))

    if t == n:
        print('Você acertou o número!')
        ac = True
        break

    elif t < n:
        print('O número secreto é maior que o número digitado!')

    else:
        print('O número secreto é menor que o número digitado!')

if not ac:
    print('Fim das tentativas!')
```

```python
# 15-Desafio final: Caixa de atendimento

s = 200

ent = 0
ret = 0
rec = 0

t_entr = 0
t_saiu = 0

cod = int(input('Digite o código: '))

while cod != 0:

    if cod == 1:
        v = float(input('Digite o valor: '))

        s = s + v
        ent = ent + 1
        t_entr = t_entr + v

    elif cod == 2:
        v = float(input('Digite o valor: '))

        if v <= s:
            s = s - v
            ret = ret + 1
            t_saiu = t_saiu + v

        else:
            print('Saldo insuficiente!')
            rec = rec + 1

    cod = int(input('Digite o código: '))

print(f'Saldo final: {s}\nQuantidade de entradas: {ent}\nQuantidade de retiradas: {ret}\nQuantidade de recusadas: {rec}\nTotal que entrou: {t_entr}\nTotal que saiu: {t_saiu}')
```
