```python
# Atividade 04 de python
# (Questão-04):Maior sequência de crescimento

numeros = []

for i in range(10):
    numeros.append(int(input(f'Digite o {i+1}º número: ')))

cont = maior = 1

for i in range(1, 10):
    if numeros[i] > numeros[i-1]:
        cont += 1
        if cont > maior:
            maior = cont
    else:
        cont = 1

print(f'\nLista: {numeros}')
print(f'Maior sequência crescente: {maior}')
```

```python
# (Questão-05):Pontos no Plano

pontos = []

for i in range(6):
    x = float(input('Digite o valor de x: '))
    y = float(input('Digite o valor de y: '))
    pontos.append((x, y))

q1 = q2 = q3 = q4 = e = 0
maior_distancia = -1
ponto_distante = None

for x, y in pontos:
    if x > 0 and y > 0:
        q1 += 1
    elif x < 0 and y > 0:
        q2 += 1
    elif x < 0 and y < 0:
        q3 += 1
    elif x > 0 and y < 0:
        q4 += 1
    else:
        e += 1

    distancia = x**2 + y**2
    if distancia > maior_distancia:
        maior_distancia = distancia
        ponto_distante = (x, y)

print('\n--- RESULTADOS ---')
print(f'1º Quadrante: {q1}')
print(f'2º Quadrante: {q2}')
print(f'3º Quadrante: {q3}')
print(f'4º Quadrante: {q4}')
print(f'Eixos / Origem: {e}')
print(f'Ponto mais distante {ponto_distante} com x² + y² = {maior_distancia}')
```

```python
#(Questão-06):Frequência e palavra mais repetida
freq = {}

for i in range(10):
    p = input(f'Digite a {i+1}ª palavra: ').strip().lower()
    if p in freq:
        freq[p] += 1
    else:
        freq[p] = 1

maior_freq = 0
for qtd in freq.values():
    if qtd > maior_freq:
        maior_freq = qtd

mais_repetidas = []
for p, qtd in freq.items():
    if qtd == maior_freq:
        mais_repetidas.append(p)

print(f'\nFrequências: {freq}')
print(f'Maior frequência: {maior_freq}')
print(f'Mais repetida(s): {", ".join(mais_repetidas)}')
```

```python
#(Questão-07):Boletim de uma turma

turma = []

for i in range(6):
    print(f'\n--- Aluno {i+1} ---')
    nome = input('Nome: ')
    n1 = float(input('Nota 1: '))
    n2 = float(input('Nota 2: '))
    n3 = float(input('Nota 3: '))
    media = (n1 + n2 + n3) / 3
    turma.append({'nome': nome, 'media': media})

soma_medias = 0
for aluno in turma:
    soma_medias += aluno['media']
media_geral = soma_medias / len(turma)

maior_media = turma[0]['media']
for aluno in turma:
    if aluno['media'] > maior_media:
        maior_media = aluno['media']

acima_media = 0
top_alunos = []

print('\n=== BOLETIM ===')
for aluno in turma:
    print(f'Aluno: {aluno["nome"]} | Média: {aluno["media"]:.2f}')
    if aluno['media'] >= media_geral:
        acima_media += 1
    if aluno['media'] == maior_media:
        top_alunos.append(aluno['nome'])

print('\n=== RESUMO ===')
print(f'Média Geral da Turma: {media_geral:.2f}')
print(f'Alunos com média >= à geral: {acima_media}')
print(f'Maior média: {maior_media:.2f} (Aluno(s): {", ".join(top_alunos)})')
```

```python
#(Questão-08):Controle de estoques com movimentações

estoque = []

for i in range(5):
    print(f"\n--- Cadastro do {i+1}º Produto ---")
    cod = int(input('Código: '))
    nom = input('Nome: ')
    preco = float(input('Preço: '))
    qtd = int(input('Quantidade: '))

    estoque.append({'c': cod, 'n': nom, 'p': preco, 'q': qtd})

for i in range(5):
    print(f"\n--- Movimentação {i+1}/5 ---")
    cod = int(input('Código do produto: '))
    prod = None

    for item in estoque:
        if item['c'] == cod:
            prod = item
            break

    if prod is not None:
        tipo = input('Tipo (E/S): ').strip().upper()
        v = int(input('Quantidade: '))

        if v > 0:
            if tipo == 'E':
                prod['q'] += v
                print("Entrada realizada com sucesso!")
            elif tipo == 'S':
                if v <= prod['q']:
                    prod['q'] -= v
                    print("Saída realizada com sucesso!")
                else:
                    print("Erro: Quantidade insuficiente em estoque.")
    else:
        print("Produto não encontrado!")

total = 0
print('\n--- ESTOQUE ATUALIZADO ---')
for item in estoque:
    subtotal = item['q'] * item['p']
    total += subtotal
    print(f'Código: {item["c"]} | Nome: {item["n"]} | Qtd: {item["q"]} | Preço: R${item["p"]:.2f}')

print(f'\nTotal investido em estoque: R${total:.2f}')
```

```python
#(Questão-09):Ranking de jogadores

j = [
    {'nome': 'Maria', 'pontuacao': [10, 20, 30]},
    {'nome': 'Abder', 'pontuacao': [15, 15, 15]},
    {'nome': 'Weydenn', 'pontuacao': [30, 20, 10]},
    {'nome': 'Thales', 'pontuacao': [5, 5, 5]},
    {'nome': 'Aurora', 'pontuacao':[6, 9, 30 ]}
]

for x in j:
    pontos = x['pontuacao']

    x['t'] = pontos[0] + pontos[1] + pontos[2]

    maior = pontos[0]
    if pontos[1] > maior: maior = pontos[1]
    if pontos[2] > maior: maior = pontos[2]

    menor = pontos[0]
    if pontos[1] < menor: menor = pontos[1]
    if pontos[2] < menor: menor = pontos[2]

    x['d'] = maior - menor

m_t = j[0]['t']
m_d = j[0]['d']

for x in j:
    if x['t'] > m_t:
        m_t = x['t']
    if x['d'] < m_d:
        m_d = x['d']

top = []
reg = []

for x in j:
    if x['t'] == m_t:
        top.append(x['nome'])

    if x['d'] == m_d:
        reg.append(x['nome'])

    print(x['nome'], x['t'], x['t'] / 3)

print(top)
print(reg)
```

```python
#(Questão-10):Agenda de contatos sem duplicidades

agenda = []

while True:
    n = input('Nome (ou FIM): ').strip()
    if n.upper() == 'FIM':
        break

    achou = False
    for item in agenda:
        if item['n'].lower() == n.lower():
            achou = True
            break

    if achou:
        print('Erro: Esse nome já está cadastrado!')
    else:
        t = input('Telefone: ').strip()
        agenda.append({'n': n, 't': t})

busca = input('\nBuscar nome (ou FIM): ').strip()
while busca.upper() != 'FIM':
    achou = False
    for item in agenda:
        if item['n'].lower() == busca.lower():
            print(f'Telefone de {item["n"]}: {item["t"]}')
            achou = True
            break

    if not achou:
        print('Nome não encontrado!')

    busca = input('Buscar nome (ou FIM): ').strip()
```

```python
#(Questão-11):Pesquisa de votos

votos = {1: 0, 2: 0, 3: 0, 4: 0}
tot_votos = 0

while True:
    v = int(input('Voto (1-4 ou 0 para sair): '))
    if v == 0:
        break
    if v in votos:
        votos[v] = votos[v] + 1
        tot_votos = tot_votos + 1
    else:
        print('Invalido')

maior = -1
vencedores = []

for op in votos:
    if votos[op] > maior:
        maior = votos[op]

for op in votos:
    if votos[op] == maior and maior > 0:
        vencedores.append(op)

print('Totais:', votos)

for op in votos:
    if tot_votos > 0:
        perc = (votos[op] / tot_votos) * 100
    else:
        perc = 0.0
    print('Opcao', op, ':', perc, '%')

print('Vencedor(es):', vencedores)
```

```python
#(Questão-12):Histórico de movimentações

estoque = {101: 50, 102: 30, 103: 10}

movs = [
    (101, 'entrada', 20),
    (102, 'saida', 10),
    (103, 'saida', 15),
    (104, 'entrada', 5),
    (101, 'saida', 10),
    (102, 'entrada', 5),
    (103, 'saida', 5),
    (102, 'saida', 40)
]

hist = []

for m in movs:
    cod = m[0]
    tipo = m[1]
    qtd = m[2]

    if cod in estoque:
        if tipo == 'entrada':
            estoque[cod] = estoque[cod] + qtd
            hist.append((cod, tipo, qtd, 'OK'))
        elif estoque[cod] >= qtd:
            estoque[cod] = estoque[cod] - qtd
            hist.append((cod, tipo, qtd, 'OK'))
        else:
            hist.append((cod, tipo, qtd, 'Sem estoque'))
    else:
        hist.append((cod, tipo, qtd, 'Nao encontrado'))

print('Estoque:', estoque)
print('Historico:', hist)
```

```python
#(Questão-13):Análise de vendas por revendedor

vendedores = [
    {'nome': 'Clara', 'vendas': [100.0, 200.0, 150.0, 50.0]},
    {'nome': 'Helenna', 'vendas': [300.0, 200.0, 100.0, 400.0]},
    {'nome': 'Cardoso', 'vendas': [50.0, 100.0, 50.0, 100.0]},
    {'nome': 'Wendel', 'vendas': [250.0, 250.0, 250.0, 250.0]},
    {'nome': 'Samuel', 'vendas': [300.0, 200.0, 400.0, 100.0]}
]

total_geral = 0.0

for v in vendedores:
    tot = 0.0
    for val in v['vendas']:
        tot = tot + val
    v['total'] = tot
    total_geral = total_geral + tot

media = total_geral / 5

maior = -1.0
top = []
acima = 0

for v in vendedores:
    if v['total'] > maior:
        maior = v['total']

for v in vendedores:
    if v['total'] == maior:
        top.append(v['nome'])
    if v['total'] > media:
        acima = acima + 1

for v in vendedores:
    print(v['nome'], ':', v['total'])

print('Total geral:', total_geral)
print('Maior total:', top)
print('Acima da media:', acima)
```

```python

#(Questão-14): Catálogo e pedidos

produtos = []

for i in range(5):
    codigo = input(f'Código: ')
    nome = input(f'Nome: ')
    preco = float(input(f'Preço: '))
    estoque = int(input(f'Estoque: '))

    produtos.append({
        'codigo': codigo,
        'nome': nome,
        'preco': preco,
        'estoque': estoque
    })

vendas = []
total = 0
unidades = 0

while True:
    codigo = input(f'\nCódigo da venda (0 para sair): ')

    if codigo == '0':
        break

    qtd = int(input(f'Quantidade: '))

    for p in produtos:
        if p['codigo'] == codigo and p['estoque'] >= qtd:
            p['estoque'] -= qtd
            valor = p['preco'] * qtd
            vendas.append((codigo, qtd, valor))
            total += valor
            unidades += qtd
            print(f'Venda realizada!')

print(f'\nESTOQUE:')
for p in produtos:
    print(f'{p["nome"]}: {p["estoque"]}')

print(f'\nUnidades vendidas: {unidades}')
print(f'Valor total: R$ {total:.2f}')

print(f'\nHISTÓRICO:')
for v in vendas:
    print(v)

print(f'\nRESUMO:')
print((len(vendas), unidades, total))
```

```python
#(Questão-15):Sistemas de vendas e relatório final

produtos = [
    {'cod': 10, 'nome': 'Celular', 'preco': 1500.0, 'qtd': 10},
    {'cod': 20, 'nome': 'Capa', 'preco': 30.0, 'qtd': 50},
    {'cod': 30, 'nome': 'Carregador', 'preco': 100.0, 'qtd': 20},
    {'cod': 40, 'nome': 'Pelicula', 'preco': 20.0, 'qtd': 30},
    {'cod': 50, 'nome': 'Fone', 'preco': 200.0, 'qtd': 15}
]
historico = []

while True:
    cod = int(input('Codigo: '))
    if cod == 0:
        break
    qtd = int(input('Quantidade: '))

    for p in produtos:
        if p['cod'] == cod:
            if p['qtd'] >= qtd:
                p['qtd'] = p['qtd'] - qtd
                historico.append((cod, p['nome'], qtd, qtd * p['preco']))
            else:
                print('Sem estoque')
            break

tot_unidades = 0
tot_valor = 0.0
top_produto = ''
maior_val = -1.0
sem_venda = []

for p in produtos:
    qtd_p = 0
    val_p = 0.0

    for h in historico:
        if h[0] == p['cod']:
            qtd_p = qtd_p + h[2]
            val_p = val_p + h[3]

    tot_unidades = tot_unidades + qtd_p
    tot_valor = tot_valor + val_p

    if val_p > maior_val:
        maior_val = val_p
        top_produto = p['nome']

    if qtd_p == 0:
        sem_venda.append(p['nome'])

resumo = (len(historico), tot_unidades, tot_valor)

print('Estoque restante:', produtos)
print('Unidades vendidas:', tot_unidades)
print('Valor total:', tot_valor)
print('Maior em vendas:', top_produto)
print('Sem vendas:', sem_venda)
print('Historico:', historico)
print('Resumo:', resumo)
```
