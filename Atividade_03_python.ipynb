```python
# 4 - Lista de números pares e ímpares
n = []
p = []
im = []

for i in range (10):
  x = int(input('Digite um número pertencente ao conjunto dos números inteiros: '))
  n.append(x)

for x in n:
   if x % 2 == 0:
     p.append(x)
   else:
     im.append(x)

print(f'Lista completa: {n}')
print(f'Lista de números pares: {p}')
print(f'Lista de números ímpares: {im}')
```

```python
# 5 - Maior e menor da lista

x = []

for i in range(8):
    n = int(input('Digite um número: '))
    x.append(n)

me = x[0]
ma = x[0]

for i in range(8):
    if x[i] > ma:
        ma = x[i]

    if x[i] < me:
        me = x[i]

pdma = x.index(ma)
pdme = x.index(me)

print(f'Lista: {x}')
print(f'Maior: {ma}')
print(f'Posição do maior: {pdma}')
print(f'Menor: {me}')
print(f'Posição do menor: {pdme}')
```

```python

# 6 - Nomes sem repetição
t = []

while True:
 n = input('Digite um nome se quiser continuarou FIM para parar: ')

 if n == 'FIM':
  break

 if n not in t:
  t.append(n)

print(f'Lista de nomes: {t}')
print(f'Quantidade de nomes diferentes: {len(t)}')
```

```python
# 9 - Dicionário de aluno
aluno = {
    'nome': input('Digite o nome do aluno(a): '),
    'turma': input('Digite a turma do aluno(a): '),
    'nota1': float(input('Digite a primeira nota do aluno(a): ')),
    'nota2': float(input('Digite a segunda nota do aluno(a): '))
}

m = (aluno['nota1']+aluno['nota2'])/2
aluno['m'] = m

if m >= 7:
  aluno['situação'] = 'Aprovado(a)'
elif m >= 5:
  aluno['situação'] = 'Recuperação'
else:
  aluno['situação'] = 'Reprovado(a)'

print(f'Nome do aluno(a): {aluno['nome']}')
print(f'Turma do aluno(a)): {aluno['turma']}')
print(f'Situação:{aluno['situação']}')
```

```python
# 12 - Estoque e atualização
estoque = {
    'Batom': 100,
    'Gloss': 20,
    'Blush': 38,
    'Rímel': 45
    }

produto = input('Digite o nome do produto: ')
quant = int(input('Digite a quantidade vendida: '))

if produto in estoque:
  if quant <= estoque[produto]:
   print('A venda pode ser feita! ')
   estoque[produto] -= quant
  else:
   print('Estoque insuficiente! ')
else:
  print('Produto não cadrastado!')

print(f'Estoque: {estoque}')
```

```python
# 13 - Cadastro de vários alunos
turma = []

for i in range(5):
 nome = input('Digite o nome do aluno(a): ')
 nota1 = float(input('Digite a primeira nota do aluno(a): '))
 nota2 = float(input('Digite a segunda nota do aluno(a): '))

 m = (nota1+nota2)/2

 aluno = {
  'nome': nome,
  'nota1': nota1,
  'nota2': nota2,
   'm'  : m
}

 turma.append(aluno)

for aluno in turma:
  print(f'Nome:{aluno['nome']}')
  print(f'Média: {aluno['m']}')
```

```python
# 14 - Catálogo com código imutável
catalogo = []

for i in range(5):
  cod = int(input('Digite o código do produto:' ))
  nome = input('Digite o nome do produt:' )
  preco = float(input('Digite o preço 10do produto:' ))

p = (cod, nome, preco)

catalogo.append(p)

bucar_p = int(input('Digite o código do produto que você deseja:' ))

encontrado = False

for p in catalogo:
  if p[0] == bucar_p:
    print(f'Nome: {p[1]}')
    print(f'Preço: {p[2]}')
    encontrado = True

if not encontrado:
 print('Código digitado não encontrado!')
```

```python
# 15 - Mini sistema de estoque

estoque = []

for i in range(5):
    cod = int(input('Digite o código do produto: '))
    nome = input('Digite o nome do produto: ')
    quant = int(input('Digite a quantidade desse produto: '))
    preco = float(input('Digite o preço desse produto: '))

    produto = {
        'Código': cod,
        'Nome': nome,
        'Quantidade': quant,
        'Preço': preco
    }

    estoque.append(produto)

valor_total = 0

for produto in estoque:
    valor_total += produto['Quantidade'] * produto['Preço']

ma_valor = 0
ma_produto = None

for produto in estoque:
    valor = produto['Quantidade'] * produto['Preço']

    if valor > ma_valor:
        ma_valor = valor
        ma_produto = produto

resumo = (len(estoque), valor_total)

print(f'Valor total: {valor_total}')
print(f'Produto de maior valor: {ma_produto}')
print(f'Resumo: {resumo}')
```
